```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x=np.array([90,4])
```

```python
weights=np.array([0.05,1])
```

```python
bias=[-5,0,5]
```

```python
for bias in bias:
    y=np.dot(x,weights)+bias
    print(y)
```

```python
if y>=0:
  prediction="pass"
else:
  prediction="fail"
print("bias=",bias,"prediction=",prediction)
```

```python
cust=np.array([[70,12],[55,8],[90,15],[40,5],[25,2],[80,10],[60,7]])
cust
```

```python
weights=np.array([0.1,2])
bias=-20
output=np.dot(cust,weights)+bias
print(output)
```

```python
for i in range(len(cust)):
  if output[i]>=0:
    print("cust",i+1,":likely to purchase")
  else:
    print("cust",i+1,":likely to not purchase")
```
